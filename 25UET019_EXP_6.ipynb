```python
tup = ()
print(tup)

# Using String
tup = ('Prem', 'shinde')
print(tup)

# Using List
li = [1, 2, 4, 5, 6]
print(tuple(li))

# Using Built-in Function
tup = tuple('sarkar')
print(tup)
```

```python
tup = tuple("DKTE")
print(tup[0])
print(tup[1:4])
print(tup[:3])

# Tuple unpacking
tup = ("All", "is", "well")

# This line unpack values of Tuple1
a, b, c = tup
print(a)
print(b)
print(c)
```
